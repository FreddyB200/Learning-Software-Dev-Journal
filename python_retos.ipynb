#EJERCICIO 1
##Invertir un String
###input: "hola" -> output: "aloh"




```python
 #metodo 1
 s = input("ingrese una palabara a reversar: ")


def reverse(string):
  reverse_string = ""

  for char in string:
    reverse_string = char + reverse_string

  print(reverse_string)

reverse(s)
```

#EJERCICIO 2
##Contar vocales
###Cuenta cuantas vocales (a,e,i,o,u) hay en un string (mayusculas y minusculas)

```python
#motodo 1
string = input("ingrese una palabra o frase a para saber cuantas vocales tiene: ")

def contar_vocales(string):
  string = string.lower()

  contador = 0
  vocales = "aeiou"

  for char in string:
    if char in vocales:
      contador += 1


  print(contador)

contar_vocales(string)
```

#COMPRENSION DE LISTAS
###forma rapida y concisa de crear listas nuevas aplicando una operacion sobre una secuencia, con o sin filtros (if).

```python
#version convencional:
nueva_lista = []
for x in range(5):
  nueva_lista.append(x**2)

print(nueva_lista)
```

```python
#con compresion de listas:
nueva_lista = [x**2 for x in range(6)] #[ expresion for elemento in iterable]
print(nueva_lista)
```

```python
#Convertir minusculas
nombres = ["Juan", "Ana", "Carlos"]
[n.lower() for n in nombres]
```

```python
#multiplicar elementos
[x*2 for x in range(10)]
```

```python
#con condicional (if)
[x for x in range(10) if x % 2 == 0]
```

```python
#con condicional ternario (if...else)
["par" if x % 2 == 0 else "impar" for  x in range (5)]
```
